```python
import pandas as pd
```

```python
df = pd.read_csv(r'Sample.csv')
df
```

```python
# 1. Write a function that returns all orders where Sales > 500.

def high_sales_orders(df):
    return df[df['Sales'] > 500]
    
print(high_sales_orders(df))
```

```python
# 2. Create a function to return orders where Category = 'Technology' and Profit > 100.

def tech_profit_orders(df):
    return df[(df['Category'] == 'Technology') & (df['Profit'] > 100)]
# result = tech_profit_orders(df)

print(tech_profit_orders(df))
```

```python
# 3. Write a function to create a new column 'Profit_Status' where Profit > 0 → 'Profit' else 'Loss'.

def add_profit_status(df):
    df['Profit_Status'] = 'Loss'
    df.loc[df['Profit'] > 0, 'Profit_Status'] = 'Profit'
    return df

print(add_profit_status(df))
```

```python
# 4. Write a function to return total sales per region.

def total_sales_region(df):
    return df.groupby('Region')['Sales'].sum()

print(total_sales_region(df))    
```

```python
# 5. Write a function to return top 5 highest profit orders.

def top5_profit_orders(df):
    return df.sort_values(by='Profit', ascending=False).head(5)

print(top5_profit_orders(df))
```

```python
# 6. Write a function to fill missing values in 'Postal Code' with 0.

def fill_postal_code(df):
    df['Postal Code'] = df['Postal Code'].fillna(0)
    return df

print(fill_postal_code(df))    
```

```python
# 7. Write a function to apply discount rule: If Sales > 1000 → 20% discount else → 10%.

def apply_discount(df):
    df['Discount'] = 0.0
    df.loc[df['Sales'] > 1000, 'Discount'] = df['Sales'] * 0.20
    df.loc[df['Sales'] <= 1000, 'Discount'] = df['Sales'] * 0.10
    return df

df = apply_discount(df)

print(df.head())
```

```python
# 8. Write a function to return all unique cities.

pd.DataFrame(df['City'].unique(), columns=['City'])
```

```python
# 9. Write a function that returns total Sales and total Profit.

def total_sales_profit(df):
    return df['Sales'].sum(), df['Profit'].sum()

sales, profit = total_sales_profit(df)

print("Total Sales:", sales)
print("Total Profit:", profit)
```

```python
# 10. Write a function to return average sales per category only for 'West' region.

def avg_sales_west(df):
    return df[df['Region'] == 'West'].groupby('Category')['Sales'].mean()

print(avg_sales_west(df))
```
