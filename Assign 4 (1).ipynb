```python
import pandas as pd
```

```python
df = pd.read_csv(r'Sample.csv')
df
```

```python
# 1. Load the Sample Superstore dataset and display the first and last 10 rows.
    
print("First 10 rows:")
print(df.head(10))

print("\nLast 10 rows:")
print(df.tail(10))
```

```python
# 2. Identify and handle missing values using at least two different techniques.

df.isnull().sum()

# Technique 1 - Imputation (Fill Missing Values)

df['Sales'] = df['Sales'].fillna(df['Sales'].mean())

df['Category'] = df['Category'].fillna(df['Category'].mode()[0])

print(df)

# Technique 2: Deletion (Drop Missing Values)

df_cleaned = df.dropna()
```

```python
# 3. Find the top 10 customers based on total sales.

top_customers = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)

print(top_customers)
```

```python
# 4. Calculate total profit by region and sort in descending order.

region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
print(region_profit)
```

```python
# 5. Create a pivot table showing sales by Category and Sub-Category.

pivot_table = pd.pivot_table(df,
                             values='Sales',
                             index='Category',
                             columns='Sub-Category',
                             aggfunc='sum')

print(pivot_table)
```

```python
# 6. Find the average discount given in each segment.

avg_discount = df.groupby('Segment')['Discount'].mean()

print(avg_discount)
```

```python
# 7. Identify the most profitable product in each category.

profitable_product = df.loc[df.groupby('Category')['Profit'].idxmax(),['Category', 'Product Name', 'Profit']]

print(profitable_product)
```

```python
# 8. Calculate month-wise sales trend using the Order Date column.

# Correctly parse date format (IMPORTANT FIX)
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

# Extract Month-Year
df['Month'] = df['Order Date'].dt.to_period('M')

# Month-wise sales trend
monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales)
```

```python
# 9. Perform groupby operation to find total quantity sold per region.

Region_Quantity = df.groupby('Region')['Quantity'].sum()

print(Region_Quantity)
```

```python
# 10. Create a new column Profit Margin and analyze its distribution.
import matplotlib.pyplot as plt

df['Profit Margin'] = df['Profit'] / df['Sales']

print(df['Profit Margin'].describe())

df['Profit Margin'].plot(kind='hist', bins=20)

plt.title("Profit Margin Distribution")
plt.xlabel("Profit Margin")
plt.ylabel("Frequency")
plt.show()      
```

```python
# 11. Filter orders where sales are above 500 and profit is negative.

filtered_orders = df[(df['Sales'] > 500) & (df['Profit'] < 0)]

print(filtered_orders)
```

```python
# 12. Find duplicate records and remove them properly.

# Find duplicate records
print("Total duplicates:", df.duplicated().sum())

# Remove duplicate records
df_cleaned = df.drop_duplicates()

# Check new shape after removing duplicates
print("Shape after removing duplicates:", df_cleaned.shape)
```

```python
# 13. Perform multi-index grouping using Region and Segment.

multi_group = df.groupby(['Region', 'Segment'])['Sales'].sum()

print(multi_group)    
```

```python
# 14. Merge two datasets (simulate Orders and Returns dataset).

orders = pd.read_csv('Sample.csv')

# Simulate Returns dataset
returns = pd.DataFrame({
    'Order ID': ['CA-2019-100006', 'CA-2019-100090', 'CA-2019-100293'],
    'Returned': ['Yes', 'Yes', 'Yes']
})

# Merge datasets on Order ID
merged_df = pd.merge(orders, returns, on='Order ID', how='left')

print(merged_df.head())
```

```python
# 15. Use apply() function to categorize orders into High, Medium, Low sales.

# Function to categorize sales
def sales_category(x):
    if x > 1000:
        return "High"
    elif x > 500:
        return "Medium"
    else:
        return "Low"

# Apply function to create new column
df['Sales Category'] = df['Sales'].apply(sales_category)

print(df[['Sales', 'Sales Category']].head())
```

```python
# 16. Identify outliers in Sales using IQR method.

# Step 1: Calculate Q1 and Q3
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

# Step 2: Calculate IQR
IQR = Q3 - Q1

# Step 3: Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Step 4: Identify outliers
outliers = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]

print(outliers)
```

```python
# 17. Rank products based on total sales within each category.

product_sales = df.groupby(['Category', 'Product Name'])['Sales'].sum().reset_index()

product_sales['Rank'] = product_sales.groupby('Category')['Sales'] \
                                    .rank(ascending=False, method='dense')

result = product_sales.sort_values(['Category', 'Rank'])

print(result)
```

```python
# 18.  Calculate cumulative sales over time.

df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

df = df.sort_values('Order Date')

df['Cumulative Sales'] = df['Sales'].cumsum()

print(df[['Order Date', 'Sales', 'Cumulative Sales']].head(10))
```

```python
# 19. Reshape data using melt() and pivot() functions.

# Reshape using melt
melted_df = pd.melt(df,
                    id_vars=['Category', 'Sub-Category'],
                    value_vars=['Sales', 'Profit'])

print(melted_df.head())

# Pivot back to wide format
pivot_df = melted_df.pivot_table(index=['Category', 'Sub-Category'],
                                 columns='variable',
                                 values='value',
                                 aggfunc='sum')

print(pivot_df)
```

```python
# 20. Create a correlation matrix and interpret key relationships.

# Select only numeric columns
numeric_df = df.select_dtypes(include=['number'])

# Create correlation matrix
correlation_matrix = numeric_df.corr()

print(correlation_matrix)
```
