```python
import pandas as pd
```

```python
df = pd.read_csv(r'Sample.csv')
df
```

```python
# 1. Write a function to return total sales only for orders with Profit > 0.

def total_sales_profit_positive(df):
    return df[df['Profit'] > 0]['Sales'].sum()

print(total_sales_profit_positive(df))
```

```python
# 2. Write a function to return the top-selling category (by Sales) in each Region.

def top_category_region(df):
    sales = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()
    return sales.loc[sales.groupby('Region')['Sales'].idxmax()]

print(top_category_region(df))
```

```python
# 3. Create a function to add a column Profit_Ratio = Profit / Sales. Handle division safely.

def add_profit_ratio(df):
    df['Profit_Ratio'] = df['Profit'] / df['Sales']
    df['Profit_Ratio'] = df['Profit_Ratio'].fillna(0)
    return df

print(add_profit_ratio(df).head())
```

```python
# 4. Write a function that takes category name and minimum sales as input and returns filtered data.

def filter_data(df, category_name, min_sales):
    return df[(df['Category'] == category_name) & (df['Sales'] >= min_sales)]

print(filter_data(df, 'Technology', 100))
print(filter_data(df, 'Furniture', 100))
print(filter_data(df, 'Office Supplies', 100))
```

```python
# 5. Write a function to find duplicate rows based on Order ID and Product Name.

def find_duplicates(df):
    return df[df.duplicated(subset=['Order ID', 'Product Name'])]

print(find_duplicates(df))
```

```python
# 6. Write a function to return monthly total sales using Order Date.

def monthly_sales(df):
    df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
    return df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

print(monthly_sales(df))
```

```python
# 7. Write a function to sort data first by Region and then by Profit descending.

def sort_region_profit(df):
    return df.sort_values(by=['Region', 'Profit'], ascending=[True, False])

print(sort_region_profit(df))
```

```python
# 8. Write a function to return rows where Sales > (mean + 2 * std).

def high_sales(df):
    threshold = df['Sales'].mean() + 2 * df['Sales'].std()
    return df[df['Sales'] > threshold]

print(high_sales(df))
```

```python
# 9. Write a function to create a pivot table with Region as index, Category as columns, and Sales
# as sum.

def sales_pivot(df):
    return df.pivot_table(index='Region', columns='Category', values='Sales', aggfunc='sum')

print(sales_pivot(df))
```

```python
# 10. Write a function to filter Furniture category, group by Sub-Category, return average Profit, and
# sort descending.

def furniture_avg_profit(df):
    return df[df['Category'] == 'Furniture'].groupby('Sub-Category')['Profit'].mean().sort_values(ascending=False)

print(furniture_avg_profit(df))
```
