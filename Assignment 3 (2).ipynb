```python
df = pd.read_csv(r'Sample.csv')
df
```

```python
# 1. Write a function to calculate 7-day rolling average of Sales.

def rolling_avg_sales(df):
    df = df.sort_values('Order Date')
    df['7_day_avg_sales'] = df['Sales'].rolling(7).mean()
    return df

print(rolling_avg_sales(df))
```

```python
# 2. Write a function to perform cohort analysis based on first purchase month.

def cohort(df):
    df['Order Date'] = pd.to_datetime(df['Order Date'])
    df['Month'] = df['Order Date'].dt.to_period('M')
    first = df.groupby('Customer ID') ['Month'].min()
    df['Cohort'] = df['Customer ID'].map(first)
    return df
print(cohort(df))
```

```python
# 3. Write a function to calculate customer lifetime value and return top 10
# customers.

def top_customer_clv(df):
    clv = df.groupby('Customer ID')['Sales'].sum()
    return clv.sort_values(ascending=False).head(10)

print(top_customer_clv(df))
```

```python
# 4. Write a function to return total profit grouped by Region and Category sorted
# descending.

def profit_region_category(df):
    return df.groupby(['Region', 'Category'])['Profit'].sum().sort_values(ascending=False)

print(profit_region_category(df))
```

```python
# 5. Write a function to assign rank to orders within each Region based on
# Sales.

def rank_orders(df):
    df['Sales_Rank'] = df.groupby('Region')['Sales'].rank(ascending=False)
    return df

print(rank_orders(df).head())
```

```python
# 6. Write a function to create a Z-score normalized Sales column. 

def zscore_sales(df):
    df['Sales_Zscore'] = (df['Sales'] - df['Sales'].mean()) / df['Sales'].std()
    return df

print(zscore_sales(df).head())
```

```python
# 7. .Write a function to replace negative profits with median profit of that
# category.

def replace_negative_profit(df):
    median_profit = df.groupby('Category')['Profit'].transform('median')
    df['Profit'] = df['Profit'].mask(df['Profit'] < 0, median_profit)
    return df

print(replace_negative_profit(df).head())
```

```python
# 8. .Write a function to create a lag feature for previous order sales per
# customer. 

def lag_sales_customer(df):
    df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
    df = df.sort_values(['Customer ID', 'Order Date'])
    
    df['Prev_Sales'] = df.groupby('Customer ID')['Sales'].shift(1)
    
    df['Prev_Sales'] = df['Prev_Sales'].fillna(0)
    
    return df

print(lag_sales_customer(df).head(10))
```

```python
# 9. Write a function to calculate cumulative sales per region over time.

def cumulative_sales_region(df):
    df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

    df = df.sort_values(['Region', 'Order Date'])
    
    df['Cumulative_Sales'] = df.groupby('Region')['Sales'].cumsum()
    
    return df

print(cumulative_sales_region(df).head(10))
```

```python
# 10. Write a function to return total sales, average discount, and max profit per
# category.

def category_summary(df):
    summary = df.groupby('Category').agg(
        Total_Sales=('Sales', 'sum'),
        Avg_Discount=('Discount', 'mean'),
        Max_Profit=('Profit', 'max')
    ).reset_index()
    return summary

print(category_summary(df))
```

```python
# 11. Write a function to extract the first word from Product Name.

def first_word_product(df):
    df['First_Word'] = df['Product Name'].str.split().str[0]
    return df

print(first_word_product(df).head())
```

```python
# 12. Write a function to label orders as High, Medium, or Loss based on Profit.

import numpy as np

def label_profit_orders(df):
    conditions = [
        df['Profit'] > 100,          # High profit
        (df['Profit'] > 0) & (df['Profit'] <= 100),  # Medium profit
        df['Profit'] <= 0            # Loss
    ]
    choices = ['High', 'Medium', 'Loss']
    df['Profit_Label'] = np.select(conditions, choices, default='Unknown')
    return df

# Example usage
print(label_profit_orders(df).head())
```

```python
# 13. Write a function to create a pivot table with Region, Ship Mode, and mean
# Profit.

def pivot_profit(df):
    pivot = pd.pivot_table(
        df,
        values='Profit',
        index=['Region', 'Ship Mode'],   
        columns='Category',              
        aggfunc='mean',                  
        fill_value=0                     
    )
    return pivot

print(pivot_profit(df))
```

```python
# 14. Write a function to filter sales between two given dates.

def filter_sales_dates(df, start_date, end_date):
    temp_df = df.copy()  # avoid changing original df
    temp_df['Order Date'] = pd.to_datetime(temp_df['Order Date'], dayfirst=True)
    return temp_df[(temp_df['Order Date'] >= start_date) & (temp_df['Order Date'] <= end_date)]

print(filter_sales_dates(df, '2019-01-03', '2019-12-30'))
```

```python
# 15. Write a function to merge Orders and Returns data and identify returned
# orders.    

returned_orders = ['US-2019-103800', 'US-2019-112326']  # example order IDs

def merge_orders_returns(df, returned_orders):
    df['Returned'] = df['Order ID'].isin(returned_orders)
    return df

merged_df = merge_orders_returns(df, returned_orders)
print(merged_df.head())
```

```python
# 16. Write a function to remove outliers using IQR method.

def remove_outliers_iqr(df, column='Sales'):
    # Calculate Q1, Q3 and IQR
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    return df[(df[column] >= lower) & (df[column] <= upper)]

filtered_df = remove_outliers_iqr(df)
print(filtered_df.head())    
```

```python
# 17. Write a function to return top 3 products by sales in each category

def top_products_by_category(df, top_n=3):
    df_sorted = df.sort_values(['Category', 'Sales'], ascending=[True, False])
    
    top_products = df_sorted.groupby('Category').head(top_n)
    
    return top_products

print(top_products_by_category(df, top_n=3))
```

```python
# 18. Write a function to resample daily sales into monthly summary. 

def monthly_sales_summary(df):
    df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
    
    df = df.set_index('Order Date')
    
    monthly_summary = df['Sales'].resample('ME').sum().reset_index()
    
    return monthly_summary

print(monthly_sales_summary(df))
```

```python
# 19. Write a function pipeline to filter Technology, remove outliers, and group by
# Sub-Category.

def tech_sales_pipeline(df):
    tech_df = df[df['Category'] == 'Technology']
    
    Q1 = tech_df['Sales'].quantile(0.25)
    Q3 = tech_df['Sales'].quantile(0.75)
    IQR = Q3 - Q1
    tech_df = tech_df[(tech_df['Sales'] >= Q1 - 1.5*IQR) & (tech_df['Sales'] <= Q3 + 1.5*IQR)]
    
    result = tech_df.groupby('Sub-Category')['Sales'].sum()
    
    return result

print(tech_sales_pipeline(df))  
```
